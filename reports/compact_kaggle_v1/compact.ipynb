```python
%pip -q install bitsandbytes
from pathlib import Path
import zlib, base64
d=Path('/kaggle/working/neurobrain_compact_v1')
d.mkdir(exist_ok=True)
(d/'train.jsonl').write_bytes(zlib.decompress(base64.b64decode('eJzdW82OHDlyvvspuH0RIFQ3bPimi6DVrGfaq9kZj2SPgdUeWJnMKk5lkmmSWaXUYIB9iL34EfwcfpN9En9fBDOrqtXzo2MbWGDVVZVJMuKLiC++4Px4M7ic7c7lmxfmzz/epNg7/Osmz7m44WZjbpoYiguFH74K+eSS6VIcTNk7k6dx7L1rjTv61oXGbUzrc/FhN/m8N3l09uBS3hgbWtP44uSp5cd3Nz9tzHnFKbv0YL0//+Gf/vI+8IsX5rUNZo6T2bt+NIMzA16N1/ls3GB9b3KcsIg1W19M3ts0YqNcdogJS4ad3WFbL98HmzO2aEN5Yd5Oyf3OfNs7m508U/cn78Naz1rTe6yC1Uo0p5gOJoa7ZUNfudm8tcPmfXgf/nXKxezjOMIWsevMMGMfeJIbSG5ww9YlfHVvgjtiX31sDvjTB6xniykeK5x82csBu5hkG13s+3i6nUYsqtYeY8gx5b0fzehLs78zX8XTs2ze7aeUWzub3sLENh3jSyx1whnNlE0TB5yci+1j0WVyxIJdcvCRDbveZbN1XazHPzl3cNj33pd8x8O93js6Ef+g7Rw/+rfJwc0xwAo2f7K58863zgVsIHQ+DTgwj7Zs9uW191e3PIDAO+4oATrJjCmOMbtPztvhyxDhGfOnCGuP/bwsmY0vd4YouvnpLz/9w49PDet52mGzRf21d7Yv+xmrhoCN4b0w/T2OOvWtRkM7JXp6caKCq43h73/9b4BM4iVGgI+/iTCnGwH565D4JpjtlGfs2wW8K6t1K/aThNfOHOFfN29t32Nd4HAjm9WNnAgAeL7Z++YAm4g58uiDbfammVKaFYKnPUyxSxawDNYGmM28dWnrbbjNZe6d2VqGyGjH5A/W5DJ1Hf6WZ/9r8iHajW4IfzQHfO/Tbbe8vMRu0oVhoxhsb45ut/Mu35k/cB+0RTbIFzjaP/+jQXxMBR/wCSSDI/4pgTi5vi4Zp2I653qaF344zubvf/2bQYrpXFME1hZvacXqJx9+tziy5hb8YOcKswKM3iEVABHyFLblik04NZKM7TM2jPyA/TIlYEPix7vriPvCt+JirmAyogB/4MU9/nV2zG+Mrj/FO7i8l3eY9UfYS+Bajvn0obOfZDwJNNfqAR/4oO/rkf8mnEBz8VJMXtLit7e3/L+vvPnapmbKkug1rWLXCKASJ2BpSwcnt7OpJTrilAD/mnNzSchQO3F7xr4srbom+FffvnpNvAGFkjaLs8MdIrrmOfje5Gbv2gnRYE0LcKcpGGcT/BXch1KDPBqf8FZClG8NNmFRfwTAkIS1BDL0M4qdoRf5o3fuP2+BwlaqUmsL3OzzRACyCMKKRAI2TLNIOAoG7zuDjWW/7eEsWEKsqog29hi9GOAdgMrcXDMI/McsNfHt2PfBufGi6GnkwHpTb5MZpuwbvNA2xSNAGGIo5nRZZg2mDYscyI6wkQTyXj4uKLrntCcpwK4koItw3p353rVBd7YWDHNRjsgTuN8ljTHhnRzMhZ8hbmErMcEbhrEzhwDTzvS1PQIvdut7X2bFKtzWIJ5Z45F8T4Fb6WORx3+PvQJHr3r34Tqov2dijy5fEJC16v2ifT+/jC6uoDkTc92DN4pTrl3xFGMe9GwjjrgkjWPsudCjkZ/3MRUTIszX73h47wBCwDzF4lZCBUAi4lgLlFkePXYn+5VwC7ZMieYVd39PRJJ+fgAQC4KLMWuPWqRTnHb7ekA1fJ62J/ihSfHUwgi59+Oo1A1L/kCKKXj3u33ByZKSU2zoZFk4YIdW14e/3nh42PZ4XTZ72+oJwfTsIEQQTnGzA7YbTTNT6mzjWkScvDOzGvQJqWmWp4FguACllqGGSgMC0t6Zt2Al+3jacAsLqJhTapUG9SQS20o+AYIgcSpkBj7HQW33v/+DVIZfY0P4atLNosIm8ujKSjeyXxwPhmv4Nj7PFDI7WInhJiFrtzn2qOPYRQbK1Fgl+bF3tyiOfAXz4NY1Fggxz062P9SAD/QNcv8Hlxqf3bM7w3gBAGex7X7aYRPMM6DKFZ19nNoNuBKXYbLGgvjWtA5VISFdIavTLmegiGH53d3DyI9Iurup0mPZellonOzYK/3X8AWXw14b2fRvjH3BAgAIF9E553cRyWdTca2nGOgPijvA4QeUh+OvFnhU9leDrWUdzZvjS2Dfo9O+pY10A4vAnbms+z1qAJeTIiBAvNenUPUBtlJtLGXVgnzOcMta9HuLENqSSBHggEebCytFtnwWder32okB1gPPwN4vVv8gMc1IxwQ1S/XKAViMYdIgEQG/Xhe5BlG/9gUoVThcaPZaR0EVwm75zoFKj6R+CFF+kJyATdCbR+SHtQt2Xecbj1/WPHePbIFnphFnZgeMuijtdCvr/YC/ycGVmssmChMJSBiTQ0XBPd0T2LQ4IVR56kuuoQ3UIJyQayRZuRrabgCoJd+RMnWefexiZrHNuAc3Qy50RQ7C9Zbsoy09MAlukBUYNcwqdYMN81qyK78jBfwjFnoYwNJlXdVutFYsJna7sLIFB5vVGjyz7PLyHDgbgyR/flW/xt4KO66yuIVf1NXvzL/nxe2LG2iQZh+ZzFYLPrY3w1pZPKiRrVSV7/v/kDfOkpKKSXmIUUrTr+hI33TEacqPiUmFZtS88otq0juBPVHHaBaoDnN2fYfEI2VK1JRVpLkkvegyHGICCGDL2IP6S/1K7hk7yG3yrtuwTXZUfwIo6eWLhlhiauamr/1viB7tf7Ktj1K/geJtL/kD5OC0R1q8F1aVUTWUGrNhDa1daHdDwapjM7PBedlFeNUbZKuxw0YRMEULvu5Aibf1Seuu2hymoCVCJAFV8Eom4yfYKx4GKxnY0Gx9YijZtJ2RqIvWXuEIyCcvNK6J6sb2A1ZjhsvcW+ekPCMx4K2WzEtwWHukTDw0eEU1DD7ywkK0RUaFEG7Gr7bTx4/49yeF3SPNHuhJtcc5SdSYbWHWw2eE+mglsJb2PF9ZW7iIrAADMVIa32lfzy08yQB9TAVbItBsYzxkLb8nJiPpLCmT4XNyqWXHnU8krbRbBSFL8HUAo7HVEE1OyfZAW6Dh5ilA6VwjDTUqNYIpq9w1CA3vWaN3qRJ1rnKLZBlgAg325/Tblz5dVL8vkt0hct4hCGJ8brazeVu825k3NuUcw0afeT01hxilbr0GIvF2+el3cYtOwnxp+22yeKMaWp74jtIRcrH87mtgz4J6vo7g6mxHGFmzYHtvj3LK5BvKpyN6UxwJsUVlupxgk8V9SGLo/3iWE0jq2co8PghrlhICHupVVDjSYmyUVWSLAS+f+l7FNEShNi4PHINFRiyx5sF/UalMaZVhC/IzzgiwfP+QTL/dS1LvJqlYhMiCHomXxOagqunUUTxOvHNBCMSiy32GcvZuUeHQ3/UgRIU9Vz4oS6ya39MWoJfK+JhixmUumbb0v5LTvbSjqnVx96J7xh8olSox+u3lc3nPxTRG99KuiLkPVcpeCM0qvw2u9e4IriW5vnUDk3c6i3DkqfjpMAV2eqBqgBuAi73ukh1YfZY2jkzHI/jxM/ZMVJnISa3ouHJEGfHkpab1EcXGwBDMGPxnjuDLVIgGi8YXv0JEDRqKytxbtKWtY8md2nkpLKNqNPwTDNSLkq0v1YBGQUro+/teraSFCqHtEp5UtZFb6kTm8wGmmOy2F4GPvGSDhJVQXim+AboIUReyzhBoGnoq2RlrYr/MDrPoZItRwZxLHBhUHdfH4hOOHsqFbpTZOKFpjwnvEg666NtqJ+z/Mjwl9VZKit9QaNjw8X4SEcG21ECkfJ9T+UyGXa1+RIYiMitTXcq2iYlCP453kpLMvR09cm7WnLZ1iz7rWpErdWci2niR6gly34oczj56rCmoDl90viK+3FlCUyHBIo/YIpdMjTAzZMRHOQK9sEKHpz4zhdpIVMJAtR8JM+y4lU5pf2CjJK7+DCaxWKra79pQdSNnMz3JDLay+1FzyrXyv+p9Fb10pwh+D4Q+LUAky12PgO+1CjWWcyyLitaQqgI6WICMtAwxj6Ted9cZDqWPkH5B77+/eSWu0+EOHwMBB9+UzCRZC0eA9VOUzhTcXuIV+xbeeq/RIQKuUI0WlEd7ijczh6Bfz8h3mjMiYQhAby0MKfNmqxTcMlvhZyjIm5UjatFFchskvjLzuXTKvrARuTOvWCypcsE3acrLaK/hzDkO1pCl1FSjCPKJCe+wJGbZjyhivQ7H0S2DxlBfvudskNO2i3F37B42FcjqRTuLBFS4E1+s7cBSZ6q0MKAhqrrjF/ZgtYOznEqLhWe+Sao0KHiHCv3NVKhwqi1E99tPw+CqwkeHblYBqYmNzA46tZ9qv+B0yQqi4R9tHtkio3kARbmh27/TsvUYBLQCPIYC9cgoguujWFgxIKRVEGAs84noODrQLFciswvgTRS7Bm0Tr8Cyg4UW/dbiF72MkC7BghSq4yjXK3+koC3gvwDFyaahIoMz0mlYZAmf0EXC71PII2h6oLrGvrVmM7a/EkeeBaPoy7bRppa7k2rC3N4LxM4t5QqTzBGaaLoP8QLbKBAWZChs2EYLTiSyZUJLHFCFm2NNitoz5wtsDFPC/y7RgVMt8MCSfqBhfhEneEBCi9Gm3nMf9pZztptfmPay+EQN/nrDYfEeVqfvP4O8Vqgg5ybRvmpzmc9ZevtgLHxBeGNYqS6J1Dm3StnVTPq0q4Yc6RN6K6RMZ0gr0wWzA1khLbJVaqN+6ovUjnytG/XzS43/t4Q/ESd6cCBApNdbaR6TPKkH8ao5Mdij32lymPBNKuAaRVhbiQgc1WptOriS1VNV3STbPScJiqhAve+V2fis7EWmSypXIqApwHCdY0Q+raNNQr2xoy+29x+FIgYekkPaB6zoQkJH6Q3rBIu5GbDZ4Uea4acA3tX6pggpzSdRF09Onxdlh+9AiFBcMV8yM4Idr318p+kJ2dFRaFrFX8o0cFRMCon6awsSud67EGVuzwFc6aZeB1tCQSXhHtxYRyIxSE8tHcMuMa+z8bj5dIKzErYKR74oHJZCuEJY8fEZRO3qdbqCymPnUy/kbTngU4y7r/ymSv2/OsdZR61LAKriTtQNZGJtlRstTgJ7tBejnlSHPeVZVmjZWs6yr4PQxhIScBPyIwPzLTyuOiRpVEmz9kZsCINjIDV+dHVu55tZbz0RswXf6rt38XLciwKj78TnsKA0OTP9iFr1Yam8InRRfSm3hXcKk1S+ileeWwSVQVKOhNQy0+TndegLH6DvULVQywu7maHe6VDKV0A4QBhlUk05em+rXQ5V3MSjchuMQa6SPxXYmuT75TbhtzO2FpYpDmwigyF2unZXtRD/AQ9QcTrM5tW393jrjkc6c8E6PkUKqMJZDUTV9IaIzML+cbmD8SxLYiRbzJFrD3ZGGxflciWy4/6l+V7gtKU656xweY6V4RQrevyzJUXhFSVeXs7448kO7rERj+Xdu1aBZFDHnRhW9kEE6LqfPbipDXDW56tyzxlLWC4zSkTqqrwuw4V5aWd+ioEulqS4IxjHmxvHBCydBHDjG14kLedhnIzgxSLLTccpkJBVwmj1NiOvkz4i6Ko623DYqtcr0KaV2itYEt4O4XWp5rrwQ1TSA4IcbkFDwR4TOxSWajrt7WFWNvcGxWDHjlmbdhtiAvVsDNuSusRXMfmPKBxfuOYg8wltglD+klyGAMuMjHY5sKg07HQ4pDj3kfXwcr8SvPkkg2Nx2mGRAmCXPqbLm5p3j18/rJqqHvK0WEURQMvorqtZHiqrVASv2CCJoL6yPvHwbuM6RKREhWSFrPcZKsWv6KtXO3nyausnlU8nj9e882cuLNVrShekixBQgvazYobcWqpixuWVpYsQev58US1MwwxKaseUhTe/eP5cp/5b25Ok1ukc0iZTlAwFswjxLm3Wgf/FjQTAjfVSphqog7y7o2pHk6ZhK59aaYY5zhNeq9NL7cPdh4b8bIGYjhSxw8MiXPHOT2Tju8vaLq4MtOr+RdPsooIhdtj8yFAJdVmOJM2gWxth/0EGCuxyp+RjZloSIoKY772lMMyOTuWBKZWZF5jWciohzERViVwlqYMbZJ4kkS3S812lMfSp0hiAei+wtDsVPnmBSsdQitXRxZGFUyacnTD3zKzENrNd/+aJdvIH+zoXSPHrkBNvC5JDpiFkvdTFiCX9BoHJY1W18Q9Hl6gkrp6I+nZ0BAe9zJJZpnkzBZDtYxbQsc29M694CRSnGJzMhaTrAUYKooqdh8gwPHFvP3q5jny+HUcxgYOuShfkI+cXJiTtK+XbvHpZRHz04DtVW58/r9LLz+H43Rm8SSpxDhMQSC03l2swe9ZpB44io9vaeC8Ivz0jXGuEZEdqb3Xq6xYPKLpFj6O2AvNIbwFEbBadZL1lgc4riVQiPXftyQX0aEhseAD6Fe8tqJOI3hX2rU/Ijf9xhXiFtwK+tj5oxJjPrkBua/jZhjeeetcKFHUCD/raSL97iuzqluU89SLeLWXiARJZ29kLE/GzqH06tFiGABXRMEgDckBGsCIecQXgRR35L4Dmtq4QzgHlNcr1dn/FOHH3ONB1cHOGtypuxad01qxWpI9subEJvdM4iFzyCfDzlNar1df45x2d5XqNjjgkIjbkt/omCQZkfBLWixBgjmQesR+dFhyBOOOn3mBQPdXA4hkhIvL0+b8EWdWnbZK4xzo7pAAdhD2s818sjeyD/7zgrDyNWJYDsnG/JP4l431maZcB8K2+eQnA8+31C2nqqtpLPXgYyRdV//8A/WjEYA==')))
(d/'development.jsonl').write_bytes(zlib.decompress(base64.b64decode('eJztWcuSG7cV3ecrkFlO0azKdjYuR4/ylK04jpWoVJYXaDTIhogGWgCanLZLVf4Ib/J7/pKcc9FNcuiRo9nPQpoH0bivc889t+eXq97mrLc2X92oH3+5StFbfHeVp1xsf7VSVyaGYkPhL78K+WCTisFPapNir/I4DN7ZVtm9a20wdq2eu1xc2I4ud2rMOJ2Lxk24IddndM44okNRbdKbktfqXzYP1hRl73CZcUVtYtraopL9MNrME7ebowXlsnIhj5sNjuLWlcp6Ujmu1TNXrPrxxd9+Wl99XKlTKPTiIpAXy2U8fvMu8MiNetNNqnUtLlO9DpNKcdJemTgm+J4sgo5BWa+bmBCSMjbZPgYnZ1qbV/DRBnXo8B+CyBZRt/DVLq4yLtVa47KL4Yte75CnL9+FY0Ju1OtOl99//W9WGobujPWej0kW8Ij6/dff8EkLE85b2ujdtquW6CbCSy6U0RW3t3OyVQ/XEm6bnTATCxLaITrm7tPRwPH7CUCK9ojHpNHw0GYMhk7JwSHZL2ZD27jHF00svAvvwkuXMuyUziKf1gWkwOCW6+vO2aST6ZzBZTHhWQltiB7u83febvFdr810fQ0EIISkc81uHzNij0Geh6temx0ubRGAQTDOIC47e4eUIx63IUqBACSyjPLJCslJo7epJnwsXUyuTOoAK3EDpJz7W7oUx22n9qhd4+1K4eiovfuZZl0evJ74FNxHg6zVC+RgUk08AJ1Wsx+UTkmHrTTCik7VkPlIigYtyPqa2PdjQPRFTFogERdZtw3iI7oNNbIwkh1DWKuvihx7BdeWQq0EZfZO9wP95MetLRqAQZcWBywVVBt2+cl1q1Oj07UKsdS27mKIabZu+8GmmFaqGdmMvQstPgkRCcjLDYUIxHO9Bp7wj1nu8JWfMxzUVHDwg0Xy29UfYaYNg0XONUBxqn7WG4ti7LXf27n+0pMVjKuK7v/YlBGZr2h9mYg6VaI49j2T/nxCd5VJsvcypsa1aHsSxVQz0+v3tejw1nryh2MTacGOOrjSyTGBiQBjwcFZ+fXYOuESIK/DsdKheQU4PeLQhsVdo7MdOaQdiaY4DDEVVLoQvaxXHsUzRDcIZNFcYCswmBcK2fhRTJAABGLiDOjFCVSaSVVzCEWyDWupnbtu7ts5w71FeYLLvZhltFt2DKJl9wW7jcXJz0z6s0gQ3c3R4gLvYZyIBywzKEbPTYacDzqhcm4glzBxEZiJg8QE9x3ovbFGg2hJFkiCxMMotE9Wt/DTJDcgmjUrRYovLOeKlWUNvislAgHqRT+4NAP7DEy1nVrYkYwA/4awBF0kS15OrDEpkNgqMbbM8gF0q7YOQ0gd5Nwdk7O1SB0o1nTg1+D6Y5OPhbjHXQwBRLJdKC4gMdNqYVwz+jImsXMi2tnV6cibtSTX13nqG6DenBHn9bUUmhGiOGi6ET1ofCwdML2SQhmQlY34dMDMwtyEP3nu2q2PEmwlEKaD5WnwtGpjT78N4Na8B0FmuQtAIMUoG/ZxykJeSHlsNULa1zgJXXb4jaIPO6EtnbzTpDNNc6g0nMtLtf79DfM794ALcg2ScipZbdl5ZiQ7jA1yAONvZLThCgBFJ0Is4RS+OeYpkFUrsgt6v+KbIVocaMkQAYQ7sqsZXUm6rWwpNBTPhid0g49xp7zb2bM5jRHJkUZvdUsE5MIWoRGb+nxEpOiJu1Jp/8hbRTfOC8G0Qj7w+DTJavFyNJX/ECUcw4wYgbWzUUyOm+lRcEl2fj9i5kUARFLuF4RM9YTGZOJv2YV0qIuHM6JAAKNh+luxf+Jqycn3s7y4Uc+jJcWBqY5qyzIgTzGnUYs92ZJ58HYvKobZRy2/vK+4jormQnb9AyrtFnoF3pGNjroGgXPCnNTGGUzq9AnxT6yvRcZdffzp419+eZKzRzn7tVsxm75VUxxVZ/2AiaFcD8GxlxaDbkJkSFZFNDxBZZB8TP0MceXh8WbWUdTEjiHoNs/92EdQiq9NZvRAeOU6VMmuDWVGuq9uf8CRv6qvUVoZyDw7e9MqDhTaAXbhbTozfUOMvruSAQrLQTiwNjFKkxLnfB63WqYZlPsEgGjYJ62pQ0y7lbpFCUYCf8rWb6pUrlTqRUrKtM3GxTGTx9k85B2r0ZYZljxt9GxAdJItppsJChSJar7iFWw8EUWw793GzoSJB2S4LXy+0T3YQXMNyDulq3wTzYfJ6euImodbrcVB+x0uz/Vkk0Tw9zrtLJ3B3Mn19IfR4TdcQISGgYxWnIYK1d5w8GASNvD276NclSq9HqzdzaIaLUdSIGxtyKglmgP0uszyg55Tb1I8LHfzIziUq1TCdB6D1XmS+ZhR7bX6jsiKc65YCUxRpQ+kXlcKiTi0o98SvfgYfdWyZoHRy4iQ/DGvmBGQ+bQL7/tKpMl2VlOIhLFvACAZKYwUj7DEvjJkYzu9hy/fsi4SemFpcT8bA5jRLfC7UhNzihGJn8m2tcllucIZYsBiDsuw0NQQQfpG0DpEw4LwOqMD6Rr1QPsy2DYeMHuu7jPtGyoP6jdTxVcb5+YRjil6Zz+TUl8TKZe0Va+jG/y5uA27dDqz98SYDzLmsz9lyw3Eb5UULL0ILFJSzd5COLcADI64crnY29rFG3uYm25ufjTHIKObGMUt6ETJhqOmxOzLs2jomXHLhQtuQY3qYSD2T3bBzVxIYBrpAsG2FarUflyMOsUulpusJcdHSHs8PcV6OZlSGZcMt0aO3OP2s9HCb+a4NZpFW/VU7R2WGj6DsgH7uJESGOuot2iYtbzSQIUcNDW9Qvn9gcvy1muRnnLjaSGa6DAKFYiH5VWKoW4VJ/V0IVmeweRBtNjGpf7UQgxHcqxPGZ71eYv9AZffx6bwwiNkzOvz0oHMZXkRLTPXxXGjusT66kinenGZjHbflafm/D/NSbWdZdzMawenr5Zl6H7XvZrA6XMxBGYNlwVduCAVPcooZV8WWeAvxbBrT3Ca96iuNm3tFtnMxYvTtY9E0LlD4I2ypPAh5JDOj6iB7SfQfAZoXskbo0whxiYnY9+AK3vRnGCab6bI90URIY3cHW2CQOMe+k8PIWO5PYbsWXvQ6uUIjxf3PGJk0zsRwdW7Fd8TxSSyan5/9WGMfJExl7KWf4HjU7EfLPbtcbMFzXJOj6nBlivt/wwN+4cCoptl7N1XX+1jCzlPFCQ1c7IJQTzsyVPhHv7Di+z1WNpcXN8j8FtkDuSIbFL0882+FPNbvY15vTz9lpuDnJOt5IGjdB3qQRu+xlTvY3OBhNv88PMnstUnfEjb4o7PBMlbmyvXS02WTPvp7HLefHxcrpcCPYHlAixnwHgLGaDlvRdVNaYnZOjmiAjRnLXm3MvDqWhrEcpY3vBs9HwVNuSHaMFd1Js08VhWmO3kkyHiwKXFp8XIeQRPJX+Q2ENdNbD/BP6pbX7732n+pYaRgHZJFti05U1osJNOos5eYRmLP+tPv+c8vt5ctF3NDW7KZ1c9UtadtorG+ijvYyLfnzp5ocWXnIvrn5zr/wNagmio')))
"""Run on Kaggle with the frozen training and development rows supplied by the notebook."""
import hashlib
import json
import math
import time
from pathlib import Path

import torch
from huggingface_hub import model_info
from peft import LoraConfig, PeftModel, get_peft_model, prepare_model_for_kbit_training
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig, set_seed

TRAIN_HASH = '06563d7ec80747609a5d4d6d4e56a7a2d4b587d31909dc1375498e3f8dd1e0cd'
DEV_HASH = 'ef479d2a43706540d4897d2929216d7b345661ca6c354c08bdc9db72855b2f32'

set_seed(17)
out = Path('/kaggle/working/neurobrain_compact_v1')
out.mkdir(exist_ok=True)
def load_rows(name, digest):
    raw = (out / name).read_bytes()
    assert hashlib.sha256(raw).hexdigest() == digest
    return [json.loads(line)['messages'] for line in raw.splitlines()]

train = load_rows('train.jsonl', TRAIN_HASH)
dev = load_rows('development.jsonl', DEV_HASH)
assert len(train) == 13 and len(dev) == 9
repo = 'Qwen/Qwen3.5-4B'
rev = model_info(repo).sha
tok = AutoTokenizer.from_pretrained(repo, revision=rev)
quant = BitsAndBytesConfig(load_in_4bit=True, bnb_4bit_quant_type='nf4', bnb_4bit_compute_dtype=torch.float16)
base = AutoModelForCausalLM.from_pretrained(repo, revision=rev, quantization_config=quant,
    device_map={'': 0}, dtype=torch.float16, attn_implementation='eager')
base = prepare_model_for_kbit_training(base, use_gradient_checkpointing=False)
def prompt(row):
    text = tok.apply_chat_template(row[:-1], tokenize=False, add_generation_prompt=True, enable_thinking=False)
    return tok.encode(text, add_special_tokens=False)

def evaluate(model, name):
    model.eval()
    answers = []
    with torch.inference_mode():
        for row in dev:
            x = torch.tensor([prompt(row)], device='cuda')
            start = time.monotonic()
            y = model.generate(x, attention_mask=torch.ones_like(x), max_new_tokens=180,
                               do_sample=False, pad_token_id=tok.eos_token_id)
            answer = tok.decode(y[0, x.shape[1]:], skip_special_tokens=True)
            answers.append({'answer': answer, 'seconds': time.monotonic()-start})
    (out / (name+'.json')).write_text(json.dumps(answers, indent=2))
    print(name, json.dumps([a['answer'] for a in answers]), flush=True)

evaluate(base, 'baseline')
model = get_peft_model(base, LoraConfig(r=8, lora_alpha=16, target_modules=['q_proj','v_proj'], task_type='CAUSAL_LM'))
model.train()
optimizer = torch.optim.AdamW((p for p in model.parameters() if p.requires_grad), lr=0.0001)
losses = []
for epoch in range(2):
    for idx in torch.randperm(len(train)).tolist():
        row = train[idx]
        prefix = prompt(row)
        target = tok.encode(row[-1]['content'], add_special_tokens=False) + [tok.eos_token_id]
        assert len(prefix)+len(target) <= 1024
        x = torch.tensor([prefix+target], device='cuda')
        labels = torch.tensor([[-100]*len(prefix)+target], device='cuda')
        optimizer.zero_grad()
        loss = model(x, labels=labels, use_cache=False).loss
        loss.backward()
        norm = torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0, error_if_nonfinite=True)
        assert math.isfinite(loss.item()) and norm.item() > 0
        optimizer.step()
        losses.append(loss.item())
        if len(losses) == 20:
            break
model.save_pretrained(out/'adapter')
base = model.unload()
model = PeftModel.from_pretrained(base, out/'adapter')
evaluate(model, 'reloaded')
print('COMPACT_DONE', rev, len(losses), round(torch.cuda.max_memory_allocated()/2**30, 2), flush=True)
(out/'run.json').write_text(json.dumps({'revision':rev,'losses':losses,'seed':17,'gpu':torch.cuda.get_device_name()}, indent=2))
```
